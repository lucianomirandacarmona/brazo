```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

np.random.seed(123)
```

```python
# 200 random ages between 18 and 80
edades = np.random.randint(18, 81, 200)

# 200 random exercise-hour values between 0 and 10
ejercicio = np.random.randint(0, 11, 200)


# --------------------------------------------------
# 2. Create labels
# --------------------------------------------------
# Rule:
# healthy = age < 60 AND exercise >= 3

saludable = (edades < 60) & (ejercicio >= 3)

# Convert True/False to 1/0
y = saludable.astype(int)


# --------------------------------------------------
# 3. Create the feature matrix X
# --------------------------------------------------
# Each row represents one person:
# [age, exercise_hours]

X = np.column_stack((edades, ejercicio))


print("X:")
print(X[:10])

print("\ny:")
print(y[:10])

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

```python
model = DecisionTreeClassifier()

model.fit(X_train, y_train)
```

```python
predicciones = model.predict(X_test)

print(predicciones)
```

```python
accuracy = accuracy_score(y_test, predicciones)

print("Accuracy:", accuracy)
```

```python
persona = np.array([[25, 5]])

resultado = model.predict(persona)

print(resultado)
```
